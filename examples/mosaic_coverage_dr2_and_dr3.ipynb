# LoTSS DR2 and DR3 Mosaic Coverage Comparison

This notebook demonstrates how to load LoTSS mosaic metadata for DR2 and DR3 and visualize their sky coverage on an Aitoff projection.

It walks through:
- importing the required helpers from `strw_lofar_data_utils`
- loading DR2 and DR3 mosaic lists from the default coverage files
- comparing the number of mosaics in each release
- plotting DR2 and DR3 coverage separately

## 1) Imported dependencies

The cell above imports the plotting helper (`Vizualizer`) and the mosaic-loading utility (`get_list_of_mosaics`).

```python
import pandas as pd

from strw_lofar_data_utils.core.vizualization import Vizualizer
from strw_lofar_data_utils.core.mosaic import get_list_of_mosaics
```

## 2) Loaded DR2 and DR3 mosaic lists

The cell above loads mosaic coverage metadata for both releases using package defaults.
If your environment variables for mosaic base directories are set, these lists are ready for plotting.

```python
dr2_mosaics = get_list_of_mosaics(mosaic_coverage_file='default', release='dr2')
dr3_mosaics = get_list_of_mosaics(mosaic_coverage_file='default', release='dr3')
```

## 3) Quick sanity check

The cell above prints the number of mosaics in DR2 and DR3.
Use this to confirm both coverage files were found and parsed correctly.

```python
print(f"Number of DR2 mosaics: {len(dr2_mosaics)}")
print(f"Number of DR3 mosaics: {len(dr3_mosaics)}")
```

## 4) DR2 coverage plot

The cell above visualizes the DR2 mosaic footprint on an Aitoff sky projection.
This helps inspect how DR2 tiles populate the northern sky.

```python
Vizualizer().vizualize_mosaic_coverage(dr2_mosaics)
```

## 5) DR3 coverage plot

The cell above shows DR3 coverage using the same projection and style,
which makes DR2 vs DR3 footprint differences easy to compare visually.

```python
Vizualizer().vizualize_mosaic_coverage(dr3_mosaics)
```
