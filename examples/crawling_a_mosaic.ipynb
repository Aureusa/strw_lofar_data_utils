## This notebook demonstrates how to crawl a DR2 Mosaic on the STRW cluster using the `strw_lofar_data_utils` package. It covers:
- Using the Crawler class to systematically extract cutouts from a mosaic image
- Configuring cutout size and stride parameters to control overlap
- Visualizing crawl coverage across the mosaic
- Comparing different stride values (no overlap vs. 50% overlap) for comprehensive coverage

Crawling is useful for various analyses, such as source detection, classification, or training machine learning models.

# Import necessary libraries and modules:

```python
from strw_lofar_data_utils.core.mosaic_crawler import Crawler
from strw_lofar_data_utils.core.vizualization import Vizualizer
```

# Craw a given mosaic

```python
mosaic_name = 'P039+31'
cutout_size = 300 # pixels
stride = 1 # stride of 0.5 means 50% overlap, stride of 1 means no overlap, etc.

dr2crawler = Crawler(mosaic_name, cutout_size, stride)
```

```python
cutouts = dr2crawler.crawl()
Vizualizer().vizualize_a_crawl(dr2crawler)
```

# Let's do a stride of 0.5, which means 50% overlap between cutouts. This is a common choice to ensure that we capture all features in the mosaic without missing anything due to the cutout boundaries.

```python
stride = 0.5
dr2crawler = Crawler(mosaic_name, cutout_size, stride)
cutouts = dr2crawler.crawl()
Vizualizer().vizualize_a_crawl(dr2crawler)
```

# Crawl a DR3 mosaic

```python
mosaic_name = 'P033+44'
cutout_size = 300 # pixels
stride = 1 # stride of 0.5 means 50% overlap, stride of 1 means no overlap, etc.

dr3crawler = Crawler(mosaic_name, cutout_size, stride, release='DR3')
```

```python
cutouts = dr3crawler.crawl()
Vizualizer().vizualize_a_crawl(dr3crawler)
```

```python
stride = 0.5
dr3crawler = Crawler(mosaic_name, cutout_size, stride, release='DR3')
cutouts = dr3crawler.crawl()
Vizualizer().vizualize_a_crawl(dr3crawler)
```
