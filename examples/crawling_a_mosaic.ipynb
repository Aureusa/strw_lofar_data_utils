## This notebook demonstrates how to craw a DR2 Mosaic on the STRW cluster using the `strw_lofar_data_utils` package. Crawling in this context means to systematically extract cutouts from a mosaic image, which can be useful for various analyses, such as source detection, classification, or training machine learning models.

# Import necessary libraries and modules:

```python
from src.core.mosaic_crawler import DR2Crawler
from src.core.vizualization import Vizualizer
```

# Craw a given mosaic

```python
mosaic_name = 'P039+31'
cutout_size = 300 # pixels
stride = 1 # stride of 0.5 means 50% overlap, stride of 1 means no overlap, etc.

crawler = DR2Crawler(mosaic_name, cutout_size, stride)
```

```python
cutouts = crawler.crawl()
Vizualizer().vizualize_a_crawl(crawler)
```

# Let's do a stride of 0.5, which means 50% overlap between cutouts. This is a common choice to ensure that we capture all features in the mosaic without missing anything due to the cutout boundaries.

```python
stride = 0.5
crawler = DR2Crawler(mosaic_name, cutout_size, stride)
cutouts = crawler.crawl()
Vizualizer().vizualize_a_crawl(crawler)
```
